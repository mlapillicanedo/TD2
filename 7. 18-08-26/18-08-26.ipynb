# Seguimos...

Demora inicial de una tarea: Hay tres parámetros para construir la **lista de ticks**:

- El período de la tarea $(T_i)$
- El peor tiempo de ejecución de la tarea $(WCET_i)$
- La **demora incicial** de la tarea $(Offset_i)$

La demora inicial de la tarea permite modificar:
- La **precedencia** de la tareas.
- El **tiempo de respuesta** a situaciones o requerimientos determinados.
- Porcentaje de **carga de CPU** para cada tick.
