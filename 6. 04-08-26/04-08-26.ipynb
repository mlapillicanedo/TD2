# Sistemas Gobernados por Tiempo (TDS)

Hay tres tipos

- Estrictos: Si no se alcanza la respuesta en el tiempo determinado el sistema **falla**

- Firme: Pueden no alcanzar el tiempo de generación de la salida ocasionalmente. La respuesta tardía **no tiene valor**.

- Flexible: Pueden no alcanzar el tiempo de generación de la salida ocasionalmente. El valor de la respuesta **decrece con el tiempo**.



El tiempo de ejecución debe ser determinista. **NO** debe ser eficiente, sino que debe responder acorde a las necesidad de la planta previendo el **peor caso posible**.
