# DMA e Introducción a la IEC61508

DMA significa acceso directo a memoria. Independientemente de si es HARVARD o VON NEUMANN, el bus de direcciones siempre lo maneja el procesador 

En general es un sistema con un único procesador, un periferico que puede copiar zonas de memoria a "espaldas" del microprocesador.
Un periferico puede ser leido y escrito por el procesador y a su vez pueden leer y/o escribir a otros perifericos.
