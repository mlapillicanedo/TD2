Se asigna un stack a cada tarea, conveniente con la arquitectura que usamos ya que tenemos dos stack pointers.

**Scheduler:** tambien llamado **planificador** en español, se encargará de ejecutar y administrar los datos según prioridad. Se puede representar como una máquina de estado de dos estados (running y not running).

**Planificación cooperativa:** (**RTC**; **R**un **t**o **C**ompletion scheduler) cada tarea se ejecuta hasta que termina y **la prioridad la tiene tarea en ejecución**. El **scheduler** es una **lista de tareas** a ejecutar.

**Round-Robin:** (**RR**, **Time Slice**). El **scheduler** tiene la capacidad de poner y sacar tareas en ejecución y a cada caso se le asigna una cantidad de "**ticks**" (tiempos) de ejecución. Las tareas tienen la **misma prioridad**

**Apropiativo:** (**Preemptive**). En este tipo de **scheduler** cada tarea tiene una prioridad determinada. El **scheduler va a ejecutar la tarea de mayor prioridad** que esté en condiciones de ejecutarse.

**Tarea bloqueada:** Tarea que **NO** ejecuta código, desaparece del **scheduler** esperando a que un **evento libere el recurso** para que la tarea esté lista.
